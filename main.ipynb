```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import json
import pandas as pd

# Initialize the WebDriver (make sure to use appropriate driver)
driver = webdriver.Chrome()

# data_dict = {
#     "Detects": {
#         "Solid": "https://www.productioninspiration.com/8/detects/solid",
#         "Liquid": "https://www.productioninspiration.com/8/detects/liquid",
#         "Gas": "https://www.productioninspiration.com/8/detects/gas",
#         "Field": "https://www.productioninspiration.com/8/detects/field"
#     }
# }

results = []
error_count = 0
count_items = 0

for effect, states in data_dict.items():
    for state, url in states.items():
        driver.get(url)
        
        # Wait for items to load
        try:
            WebDriverWait(driver, 4).until(
                EC.presence_of_element_located((By.CLASS_NAME, "item"))
            )
        except:
            print(f"Timeout waiting for items to load on {url}")
            error_count += 1
            continue

        # Find all item containers
        items = driver.find_elements(By.CLASS_NAME, "item")
        
        for item in items:

            try:
                # Extract title
                title = item.find_element(By.TAG_NAME, "h3").text
                
                # Extract description (first p tag that's not example)
                description = item.find_element(
                    By.XPATH, ".//p[not(contains(@class, 'item-example'))]"
                ).text
                
                # Extract example if exists
                try:
                    example = item.find_element(
                        By.CLASS_NAME, "item-example"
                    ).find_element(By.TAG_NAME, "small").text
                except:
                    example = None
                
                # Add to results
                results.append({
                    "Effect": effect,
                    "State": state,
                    "Title": title,
                    "URL": url,
                    "Description": description,
                    "Example": example
                })
                count_items += 1
            except Exception as e:
                print(f"Error processing item: {e}")
                error_count += 1
                continue

driver.quit()

df = pd.DataFrame(results)
df.to_csv('productioninspiration_data.csv', index=False)

print(f"Data successfully saved to results.json")
print(f"Total items processed: {count_items}")
print(f"Errors encountered: {error_count}")
```
